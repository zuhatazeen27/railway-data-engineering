```python
from pyspark.sql.functions import current_timestamp, lit

catalog = "railway_data_engineering"
bronze_schema = "bronze"

raw_path = "/Volumes/railway_data_engineering/bronze/railway_raw/Railway_info.csv"

bronze_table = "railway_data_engineering.bronze.railway_bronze"

print("Raw file:", raw_path)
print("Bronze table:", bronze_table)
```

```python
df_bronze = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(raw_path)
display(df_bronze)
```

```python
df_bronze.printSchema()
```

```python
print("Total Bronze Records:", df_bronze.count())
```

```python
display(df_bronze.limit(10))
```

```python
# Add Bronze ingestion metadata
df_bronze = df_bronze \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("source_file", lit("Railway_info.csv"))

display(df_bronze.limit(10))
```

```python
# Write data into the Bronze Delta table
df_bronze.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("railway_data_engineering.bronze.railway_bronze")

print("✅ Bronze Delta table created successfully!")
```

```python
# Verify Bronze Delta table
bronze_check = spark.table(
    "railway_data_engineering.bronze.railway_bronze"
)

print("Bronze record count:", bronze_check.count())

display(bronze_check.limit(10))
```
